```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load image
img = cv2.imread("sig.jpg")

def clean_process(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Step 2. Denoise to remove small paper texture
    denoised = cv2.fastNlMeansDenoising(gray, None, h=20, templateWindowSize=7, searchWindowSize=21)

    # Step 3. Adaptive threshold to separate ink from background
    thresh = cv2.adaptiveThreshold(
        denoised, 255,
        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
        cv2.THRESH_BINARY_INV,
        15, 10
    )

    # Step 4. Morphological opening to remove small noise dots
    kernel = np.ones((2,2), np.uint8)
    clean = cv2.morphologyEx(thresh, cv2.MORPH_OPEN, kernel, iterations=1)

    # Step 5. Optional dilation to make the signature slightly thicker
    clean = cv2.dilate(clean, kernel, iterations=1)

    # Step 6. Invert back to black ink on white background
    clean_inv = cv2.bitwise_not(clean)

    # Step 7. Enhance brightness / flatten background
    # Use contrast stretching to make whites whiter and blacks darker
    clean_inv = cv2.normalize(clean_inv, None, 220, 255, cv2.NORM_MINMAX)

    # Step 8. Optional slight smoothing to remove any rough edges
    final = cv2.GaussianBlur(clean_inv, (3,3), 0)

    return final

# Step 7. Save result
# cv2.imwrite("cleaned_document.png", final)
```

```python
scanned = clean_process(img)

plt.figure(figsize=(10,5))
plt.imshow(scanned, cmap='gray')
plt.title("Clean Scanned Signature")
plt.axis('off')
plt.show()
```
