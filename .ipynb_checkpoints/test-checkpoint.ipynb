```python
import os
import cv2
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.data import Data
from torch_geometric.nn import GCNConv, VGAE
from torch_geometric.nn import global_mean_pool
from torchvision import transforms
from PIL import Image
import numpy as np
import pandas as pd
from tqdm import tqdm
import matplotlib.pyplot as plt
import seaborn as sns

from lib.lib import image_to_graph
```

```python
device = torch.device('cuda:0' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
```

```python
class GNNEncoder(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels, latent_dim):
        super(GNNEncoder, self).__init__()
        self.conv1 = GCNConv(in_channels, hidden_channels)
        self.conv_mu = GCNConv(hidden_channels, latent_dim)
        self.conv_logvar = GCNConv(hidden_channels, latent_dim)

    def forward(self, x, edge_index):
        # Step 1: Aggregate node features from neighbors
        x = F.relu(self.conv1(x, edge_index))

        # Step 2: Output mean and log variance
        mu = self.conv_mu(x, edge_index)
        logvar = self.conv_logvar(x, edge_index)

        return mu, logvar

class SiameseNetwork(nn.Module):
    def __init__(self, fe_model, latent_dim):
        super(SiameseNetwork, self).__init__()
        self.encoder = fe_model
        self.embedding_dim = latent_dim
        
        self.projector = nn.Sequential(
            nn.Linear(self.embedding_dim, 128),
            nn.ReLU(inplace=True),
            nn.Dropout(p=0.5),

            nn.Linear(128, 64),
            nn.ReLU(inplace=True),
            nn.Dropout(p=0.5),

            nn.Linear(64, 2)
        )

    def forward_once(self, x, edge_index, batch):
        mu, _ = self.encoder(x, edge_index)
        graph_emb = global_mean_pool(mu, batch) 
        # x = torch.flatten(x, 1)
        # return x
        return graph_emb

    def forward(self, x1, x2,
               edge_index1, edge_index2,
               batch):
        emb1 = self.forward_once(x1, edge_index1, batch)
        emb2 = self.forward_once(x2, edge_index2, batch)
        
        emb1 = F.normalize(emb1, p=2, dim=1)
        emb2 = F.normalize(emb2, p=2, dim=1)

        # Combine embeddings (abs difference works well for verification)
        combined = torch.abs(emb1 - emb2)

        # Predict same/forged
        out = self.projector(combined)
        return out

def load_model(vgae_path, siamese_path):
    """Load the trained VGAE and Siamese models"""
    input_dim = 3
    hidden_dim = 64
    latent_dim = 128
    
    # Load VGAE encoder
    vgae = VGAE(GNNEncoder(in_channels=input_dim, 
                          hidden_channels=hidden_dim, 
                          latent_dim=latent_dim)).to(device)
    vgae.load_state_dict(torch.load(vgae_path, map_location=device))
    vgae.eval()
    
    # Load Siamese model
    model = SiameseNetwork(vgae, latent_dim=128).to(device)
    model.load_state_dict(torch.load(siamese_path, map_location=device))
    model.eval()
    
    return model
```

```python
class SignaturePreprocessor:
    """
    Advanced preprocessing pipeline for signature images
    Cleans and normalizes signatures for better accuracy
    """
    
    def __init__(self, target_size=(32, 32)):
        self.target_size = target_size
    
    def get_transform(self):
        """Return the transformation pipeline"""
        return transforms.Compose([
            transforms.Grayscale(num_output_channels=1),
            transforms.Resize(self.target_size),
            transforms.ToTensor(),
        ])

def clean_process(img):
    # Step 1. Convert to grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Step 2. Denoise to remove paper grain or texture
    denoised = cv2.fastNlMeansDenoising(gray, None, h=15)

    # Step 3. Create a mask for the ink (dark regions)
    # Invert so that ink is white and background is black
    _, mask = cv2.threshold(denoised, 0, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)

    # Step 4. Invert mask back so signature is black (0) and background is white (255)
    inv_mask = cv2.bitwise_not(mask)

    # Step 5. Create a pure white image (same size as input)
    white_bg = np.full_like(img, 255)

    # Step 6. Combine white background with original image using mask
    # Keeps original ink, replaces background with white
    result = np.where(mask[..., None] == 255, img, white_bg)

    return result

def preprocess_image(image_path, transform_fn):
    try:
        # Load and convert image
        image = Image.open(image_path).convert('RGB')
        
        # Apply transformations (resize is already in transform_fn)
        image_tensor = transform_fn(image)
        
        # Convert to graph (assuming this function exists elsewhere)
        graph = image_to_graph(image_tensor)
        
        return graph
    except Exception as e:
        print(f"Error processing {image_path}: {e}")
        return None
```

```python
def compare_single_pair(graph1, graph2, model, device):
    """Compare two signature graphs"""
    # Add batch dimension
    graph1.batch = torch.zeros(graph1.num_nodes, dtype=torch.long, device=device)
    graph2.batch = torch.zeros(graph2.num_nodes, dtype=torch.long, device=device)
    
    with torch.no_grad():
        output = model(
            graph1.x.to(device),
            graph2.x.to(device),
            graph1.edge_index.to(device),
            graph2.edge_index.to(device),
            graph1.batch
        )
        
        probs = torch.softmax(output, dim=1)
        pred = torch.argmax(output, dim=1).item()
        confidence_genuine = probs[0][1].item()  # Probability of being genuine
        confidence_forged = probs[0][0].item()   # Probability of being forged
    
    return pred, confidence_genuine, confidence_forged

def compare_one_to_many(input_image_path, test_folder, model, threshold=0.5):
    
    transform_fn = SignaturePreprocessor((32,32))
    transform_fn = transform_fn.get_transform()
    
    input_graph = preprocess_image(input_image_path, transform_fn)
    if input_graph is None:
        print("Failed to load input image!")
        return None
    
    # Get all test images
    test_images = [f for f in os.listdir(test_folder) 
                   if f.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp'))]
    
    if not test_images:
        print(f"No images found in {test_folder}")
        return None
    
    print(f"Found {len(test_images)} test signatures to compare\n")
    
    results = []
    
    # Compare with each test image
    for test_img in tqdm(test_images, desc="Comparing signatures"):
        test_path = os.path.join(test_folder, test_img)
        test_graph = preprocess_image(test_path, transform_fn)  # Don't save all test images
        
        if test_graph is None:
            continue
        
        pred, conf_genuine, conf_forged = compare_single_pair(
            input_graph, test_graph, model, device
        )
        
        result = {
            'test_image': test_img,
            'prediction': 'GENUINE' if pred == 1 else 'FORGED',
            'confidence_genuine': conf_genuine,
            'confidence_forged': conf_forged,
            'match': conf_genuine >= threshold
        }
        
        results.append(result)
    
    # Create DataFrame
    df_results = pd.DataFrame(results)
    
    # Sort by confidence (descending)
    df_results = df_results.sort_values('confidence_genuine', ascending=False)
    df_results = df_results.reset_index(drop=True)
    df_results.index = df_results.index + 1  # Start index from 1
    
    return df_results
```

```python
def print_results(df_results, input_image_name, threshold=0.5):
    """Print formatted results"""
    print(f"\n{'='*60}")
    print(f"COMPARISON RESULTS FOR: {input_image_name}")
    print(f"{'='*60}")
    print(f"Threshold: {threshold:.2f}")
    print(f"Total comparisons: {len(df_results)}")
    
    genuine_matches = df_results[df_results['match'] == True]
    forged_matches = df_results[df_results['match'] == False]
    
    print(f"Genuine matches: {len(genuine_matches)}")
    print(f"Forged/Different: {len(forged_matches)}")
    print(f"{'='*60}\n")
    
    # Top 10 results
    print("Top 10 Most Similar Signatures:")
    print("-" * 60)
    for idx, row in df_results.head(10).iterrows():
        status = "HIGH CONFIDENCE" if row['match'] else "LOW CONFIDENCE"
        print(f"{idx}. {row['test_image']:<30} | "
              f"Confidence: {row['confidence_genuine']:.4f} | {status}")
    
    print("\n")
```

```python
def main(input_image, test_folder, vgae_model_path, siamese_model_path, 
         threshold=0.5):
    
    # Load model
    print("Loading models...")
    model = load_model(vgae_model_path, siamese_model_path)
    print("Models loaded successfully!\n")
    
    # Compare signatures
    results = compare_one_to_many(
        input_image, test_folder, model, threshold
    )
    
    if results is None:
        print("Comparison failed!")
        return None
    
    # Display results
    input_name = os.path.basename(input_image)
    print_results(results, input_name, threshold)
    
    return results
```

```python
# Configuration
INPUT_IMAGE = "input_data/my_signature.png"
TEST_FOLDER = "test_data"
VGAE_MODEL = "VGAE_Model.pt"
SIAMESE_MODEL = "Siamese_Network_Model.pt"
THRESHOLD = 0.95  # Adjust based on your needs

# ============================================
# OPTION 1: Full Pipeline with Advanced Preprocessing (RECOMMENDED)
# ============================================
print("\n" + "="*60)
print("RUNNING WITH ADVANCED PREPROCESSING")
print("="*60)

results = main(
    input_image=INPUT_IMAGE,
    test_folder=TEST_FOLDER,
    vgae_model_path=VGAE_MODEL,
    siamese_model_path=SIAMESE_MODEL,
    threshold=THRESHOLD
)

def compute_confidence(distance, alpha=1.5):
    """Convert Euclidean distance to confidence [0, 1]."""
    return float(np.exp(-alpha * distance))

# Additional analysis
if results is not None:
    print("\n" + "="*60)
    print("STATISTICAL SUMMARY")
    print("="*60)
    print(f"Mean confidence: {results['confidence_genuine'].mean():.4f}")
    print(f"Std deviation: {results['confidence_genuine'].std():.4f}")
    print(f"Min confidence: {results['confidence_genuine'].min():.4f}")
    print(f"Max confidence: {results['confidence_genuine'].max():.4f}")
    print("="*60)
    
    # Show confidence distribution
    print("\nConfidence Distribution:")
    ave_confidence_g = compute_confidence(results['confidence_genuine'].mean())
    ave_confidence_f = compute_confidence(results['confidence_forged'].mean())
    print(f"Average Confidence Genuine: {ave_confidence_g:.2}")
    print(f"Average Confidence Forged: {ave_confidence_f:.2}")
    print("FORGED" if ave_confidence < threshold else "GENUINE")
```
